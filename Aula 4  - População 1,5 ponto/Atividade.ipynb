```python
path = r"H:\Python\3°etapa\PY-MA\Aula 4  - População 1,5 ponto\CD2022_Populacao_2010_Compatibilizada_20231222.xlsx"
```

```python
tabela = pd.read_excel(path, header=None)
```

```python
tabela
```

```python
tabela_tratada = tabela.iloc[3:5573, 1:8].copy()
```

```python
tabela_tratada
```

```python
tabela_tratada.columns = [
    "UF",
    "COD_UF",
    "COD_MUNIC",
    "Municipio",
    "População Município 2010",
    "População 2010 (Alterações de Limites até 2022)",
    "População Censo 2022"
]
```

```python
estado = ["AC", "AL", "AM", "AP", "BA", "CE", "DF", "ES", "GO", "MA", "MG", "MS", "MT", "PA", "PB", "PE", "PI", "PR", "RJ", "RN", "RO", "RR", "RS", "SC", "SE", "SP", "TO"]

estados = tabela_tratada[tabela_tratada["UF"].isin(estado)]
```

```python
estados["UF"].unique()
```

```python
pop_estado = (
    estados.groupby(["UF"], as_index=False)
      .agg(
          populacao_2010=("População Município 2010", "sum"),
          populacao_2022=("População Censo 2022", "sum"),
      )
)
```

```python
pop_estado.dtypes
```

```python

pop_estado["populacao_2010"] = pop_estado["populacao_2010"].astype(float).dropna()
pop_estado["populacao_2022"] = pop_estado["populacao_2022"].astype(float).dropna() 
```

```python
pop_estado["DIF"] = pop_estado["populacao_2022"] - pop_estado["populacao_2010"]
```

```python
pop_estado
```

```python
pop_estado.to_csv(r"H:\Python\3°etapa\PY-MA\Aula 4  - População 1,5 ponto.csv")
```
